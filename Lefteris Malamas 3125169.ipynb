## Άσκηση 1

Ένα αντικείμενο έχει σχήμα τετράγωνης πλάκας με μία διαμπερή κυκλική τρύπα στο κέντρo.
Γράψτε μια συνάρτηση με το όνομα `myfunction1`, η οποία δέχεται ως παράμετρο την πλευρά της πλάκας $L$ και υπολογίζει τον λόγο επιφάνεια προς όγκο $\frac{A}V$ του αντικειμένου.
Χρησιμοποιείστε την σταθερά $π$ της μονάδας κώδικα numpy.
Για τo πάχος της πλάκας $w$ και την διάμετρο της τρύπας χρησιμοποιείστε τις παρακάτω σχέσεις:

$$
w=0.005\cdot (1+\frac{c}{b+10})
$$
$$
d=0.1\cdot L\cdot (1+\frac{c}{a+10})
$$

όπου $a,b,c$ είναι τα 3 τελευταία ψηφία του αριθμού μητρώου σας ($c$=τελευταίο).
Η συνάρτηση πρέπει να επιστρέφει απροσδιόριστο, όταν οι τιμές εισόδου δεν είναι έγκυρες.

Ελέγξτε τα αποτελέσματα της συνάρτησης για $L=0.0$, $L=0.05$ και $L=0.20$.

```python
def myfunction1(L: np.float64) -> np.float64:
    if L <= 0:
        return np.float64(np.nan)

    # Για ΑΜ = 3125169 έχουμε a = 1, b = 6, c = 9
    w = 0.005 * (1 + 9 / 16)
    d = 0.1 * L * (1 + 9 / 11)

    if d >= L:
        return np.float64(np.nan)

    A = 2 * (L**2 - np.pi * (d/2)**2) + 4 * L * w + np.pi * d * w
    V = (L**2 - np.pi * (d/2)**2) * w

    return np.float64(A / V)
```

```python
# Πρόχειρο άσκησης 1
# - Σε αυτό το κελί μπορείτε να τυπώσετε μεταβλητές για επαλήθευση. Δεν λαμβάνεται υπόψη στην βαθμολόγηση.

print(myfunction1(0.0), myfunction1(0.05), myfunction1(0.20))
```

## Άσκηση 2

Γράψτε μια συνάρτηση `myfunction2`, η οποία δέχεται ως παράμετρο ένα διάνυσμα $P_1$ τριών στοιχείων (μονοδιάστατος πίνακας numpy.ndarray με συντεταγμένες $x_1$,$y_1$ και $z_1$) και επιστρέφει την γωνία $\theta$ που σχηματίζει το διάνυσμα $P_1$ με το διάνυσμα:

$$
P_2=\begin{bmatrix} a+1\\ b \\ c \end{bmatrix}
$$

οπου $a,b,c$ είναι τα 3 τελευταία ψηφία του αριθμού μητρώου σας ($c$=τελευταίο).

Για τον υπολογισμό χρησιμοποιείστε το τύπο του εσωτερικού γινόμενου:

$$
\cos \theta =\frac{P_1\cdot P_2}{|P_1|\cdot|P_2|}
$$

όπου $P_1\cdot P_2=x_1x_2+y_1y_2+z_1z_2$  το εσωτερικό γινόμενο και $|P_1|,|P_2|$ τα μέτρα των διανυσμάτων.

Η γωνία $\theta$ να επιστραφεί σε μοίρες στο πεδίο $[0,180]$.

Ελέγξτε τα αποτελέσματα της συνάρτησης για $P_1=\begin{bmatrix} 1\\ 0\\0\end{bmatrix}$ και $P_1=\begin{bmatrix} 5\\ 0\\ 10\end{bmatrix}$.

```python
def myfunction2(P1: np.ndarray) -> np.float64:
    # Για ΑΜ = 3125169 έχουμε P2 = [2, 6, 9]
    P2 = np.array([2.0, 6.0, 9.0])

    if np.linalg.norm(P1) == 0 or np.linalg.norm(P2) == 0:
        return np.float64(np.nan)

    cos_theta = np.dot(P1, P2) / (np.linalg.norm(P1) * np.linalg.norm(P2))
    cos_theta = np.clip(cos_theta, -1.0, 1.0)

    theta = np.arccos(cos_theta)
    return np.float64(np.degrees(theta))
```

```python
# Πρόχειρο άσκησης 2
# - Σε αυτό το κελί μπορείτε να τυπώσετε μεταβλητές για επαλήθευση. Δεν λαμβάνεται υπόψη στην βαθμολόγηση.

print(myfunction2(np.array([1., 0., 0.])), myfunction2(np.array([5., 0., 10])))
```
